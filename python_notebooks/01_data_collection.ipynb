##  Data collection process:

In this phase we collect the data we need to conduct the project:
- Nba teams and players stats (usual and advanced) are retrived via nba_api library
- To retrived information about the players salary we use webscrapping on Espn site using requests and BeautifulSoup libraries


### Libraries

```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup as bs
import lxml
from nba_api.stats.static import teams
from nba_api.stats.endpoints import leaguedashplayerstats 
from nba_api.stats.endpoints import LeagueDashTeamStats
```

### 1.Collecting teams dimension table

```python
# Setting up the custom header
custom_headers = {
    'Host': 'stats.nba.com',
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36',
    'Accept': 'application/json, text/plain, */*',
    'Accept-Language': 'en-US,en;q=0.5',
    'Referer': 'https://www.nba.com/',
    'Connection': 'keep-alive'
}
```

```python
# Getting the teams static (general) infos with nba_api

# get teams return a list of 30 nba teams dictionnary we convert this list into a useable dataframe 

teams_data=pd.DataFrame.from_dict(teams.get_teams())

display(teams_data.info())

# Narrow it down to essential info before storing:
teams_info=teams_data[['id','full_name','abbreviation']]

# store in data_raw:
teams_info.to_csv('data_raw/teams_info.csv',index=False)
```

### 2.Team performance data 

- perGame basic stats and advanced stats

```python
# Getting the teams pergame  stats for the 2025-26 season

teamstats=LeagueDashTeamStats(season='2025-26',
                              season_type_all_star='Regular Season',
                              per_mode_detailed='PerGame',
                              headers=custom_headers
                              
                              )


team_pergamestats=teamstats.get_data_frames()[0]

display(team_pergamestats.info())

# Store the team pergame stats in data_raw:
team_pergamestats.to_csv('data_raw/team_pergamestats.csv',index=False)

```

```python
# Getting the teams advanced stats per 100 possessions for the 2025-26 season

teamstats2=LeagueDashTeamStats(season='2025-26',
                              season_type_all_star='Regular Season',
                              per_mode_detailed='Per100Possessions',
                              measure_type_detailed_defense='Advanced',
                              headers=custom_headers
                              
                              )

advanced_team_stats=teamstats2.get_data_frames()[0]


display(advanced_team_stats.info())

# Store the team advanced stats in data_raw:
advanced_team_stats.to_csv('data_raw/advanced_team_stats.csv',index=False)
```

## 4.Salaries data

```python

# Setting up the custom header for salaries data collection from ESPN:


salaries_header= {
    'Host': 'www.espn.com',
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    "Accept-Encoding": "gzip, deflate, br",
    'Referer': "https://www.espn.com/nba/salaries",
    'Connection': 'keep-alive'
}
```

```python
# Scrapping the initial page:
first_page_url='https://www.espn.com/nba/salaries/_/year/2026/seasontype/4'

r_first_page=requests.get(first_page_url,headers=salaries_header)
soup_first_page=bs(r_first_page.content,features='lxml')

table1=soup_first_page.find('table')


# extract column headers
column_headers=table1.find(class_='colhead').find_all('td')

column_names=[c.string for c in column_headers]


# extract the rows and store them in a list of lists
table_rows=[]
rows_first_page=table1.find_all('tr',class_=["oddrow","evenrow"]) # we only extract the rows with data, not the header or footer rows
rows_first_page
for tr in rows_first_page:
    td=tr.find_all('td')
    row=[str(i.text.strip()) for i in td]
    table_rows.append(row)

```

```python
# function to scrap the salaries data from the ESPN website for a given page number:

def scrap_salaries_page(url,headers=None):
    """
    Scrapes salary data from a specific page of the ESPN NBA salaries website.
    """
    if headers is None:
        headers=salaries_header
    r=requests.get(url,headers=headers)
    soup=bs(r.content,features='lxml')
    table=soup.find('table')
    
    # extract the rows and store them in a list of lists
    table_rows=[]
    
    rows=table.find_all('tr',class_=["oddrow","evenrow"])
    for tr in rows:
        td=tr.find_all('td')
        row=[str(i.text.strip()) for i in td]
        table_rows.append(row)
    
    return table_rows


for page in range(2,13):
    url=f"https://www.espn.com/nba/salaries/_/year/2026/page/{page}/seasontype/4"
    page_rows=scrap_salaries_page(url,headers=salaries_header)
    table_rows+=page_rows
```

```python
# Convert the list of lists into a Dataframe and store it :

salaries_df=pd.DataFrame(table_rows,columns=column_names)

display(salaries_df.info())

display(salaries_df.head())

# Store the salaries data in data_raw:
salaries_df.to_csv('data_raw/players_salaries.csv',index=False)
```
